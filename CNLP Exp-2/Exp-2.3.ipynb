# Experiment-2: Basic Text Preprocessing

## Experiment 2.3: Write a Python program to tokenize a text and remove commonly occurring stop words such as is, the, a, an, in, of, and, to, etc. Display the original and filtered tokens. (Also, display Stop Words Separately).

```python
import nltk
nltk.download('punkt_tab', quiet=True)
nltk.download('stopwords', quiet=True)

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
```

```python
file_path = "2.3_clean_data.txt"

try:
    with open(file_path, 'r') as file:
        content = file.read()
        print(content)
except FileNotFoundError:
    print(f"Error: The file at {file_path} was not found.")
except Exception as e:
    print(f"An error occurred: {e}")
```

## 1. Tokenize the Text

```python
tokens = word_tokenize(content)

print("Number of original tokens: ", len(tokens))
print("Original Tokens:")
print(tokens)
```

## 2. Remove Stop Words

```python
stop_words = set(stopwords.words('english'))

filtered_tokens = [word for word in tokens if word.lower() not in stop_words]

print("Number of filtered tokens: ", len(filtered_tokens))
print("Filtered Tokens (without stop words):")
print(filtered_tokens)
```

## 3. Display Stop Words Found in the Text

```python
stop_words_found = [word for word in tokens if word.lower() in stop_words]

print("Number of stop words found: ", len(stop_words_found))
print("Stop Words Found in Text:")
print(stop_words_found)
```
