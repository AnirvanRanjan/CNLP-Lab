#  Experiment-2: Basic Text Preprocessing

## Experiment 2.1:  Write a Python program to perform basic text preprocessing on a given data. The program should:

```python
from collections import defaultdict
```

```python
file_path = "2.1_text_data.txt"

try:
    with open(file_path, 'r') as file:
        content = file.read()
        print(content)
except FileNotFoundError:
    print(f"Error: The file at {file_path} was not found.")
except Exception as e:
    print(f"An error occurred: {e}")
```

## 1. Convert text to lowercase. (Also display how many uppercase?)


```python
upper_cnt_ind = defaultdict(int)
for x in range(len(content)):
    if ord(content[x]) >= 65 and ord(content[x]) <= 90:
        upper_cnt_ind[x] = content[x]

print("Number of uppercase letters: ", len(upper_cnt_ind))

content_chars = list(content)

for index in upper_cnt_ind.keys():
    content_chars[index] = content_chars[index].lower()  

content = "".join(content_chars)
```

```python
content
```

## 2. Remove punctuation.        (Also display how many different types of punctuations?)


```python
hash_map = defaultdict(int)

for x in content:
    if not x.isalnum() and x != " " and x != "\n":
        hash_map[x] += 1
print("Number of different punctuations")
print(len(hash_map))
print("Each punctuation and their frequency is given below: ")

for x,y in hash_map.items():
    print(x, ": ",y)
```

```python
content = "".join(x for x in content if x.isalnum() or x.isspace())
content
```

## 3. Remove numbers.            (Also display how many numbers?)


```python
import re

numbers = re.findall(r"\d+", content)

print("Total numbers present: ", len(numbers))
print("Numbers found: ", numbers)

hash_map_nums = defaultdict(int)
num_set = set(list(map(str, range(10))))
fin_text = []

for x in range(len(content)):
    if content[x] in num_set:
        hash_map_nums[content[x]] += 1
    else:
        fin_text.append(content[x])

cleaned_string = "".join(fin_text)

print("\nNumber of different digits")
print(len(hash_map_nums))
print("Each digit and their frequency is given below: ")

for x, y in hash_map_nums.items():
    print(x, ": ", y)

print("\nCleaned Text:")
print(cleaned_string)
content = cleaned_string
```

```python
content
```

## 4. Remove extra whitespace. (How much extra whitespace is present?)

```python
stack = []
extra = 0

for x in range(len(content)):
    if x > 0 and stack[-1] == " " and content[x] == " ":
        extra += 1
    else:
        stack.append(content[x])

cleaned = "\n".join(line.strip() for line in "".join(stack).split("\n"))
extra += len("".join(stack)) - len(cleaned)

print("Amount of extra whitespace removed: ", extra)

content = cleaned
content
```
