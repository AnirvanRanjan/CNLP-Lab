## Experiment 2.2: Write a Python program to perform word tokenization and sentence tokenization on a given text. Display the resulting words and sentences separately.

### Three separate code: using NLTK, using spaCy and without using NLTK/spacy

```python
# !pip install NLTK spaCy
import nltk
import spacy
import re
```

```python
file_path = "2.2_tokenization_data.txt"

try:
    with open(file_path, 'r') as file:
        content = file.read()
        print(content)
except FileNotFoundError:
    print(f"Error: The file at {file_path} was not found.")
except Exception as e:
    print(f"An error occurred: {e}")
```

## NLTK Word Tokenization

```python
nltk.word_tokenize(content)
```

## NLTK Sentence Tokenization

```python
nltk.sent_tokenize(content)
```

## SpaCy Word Tokenization

```python
nlp = spacy.load('en_core_web_sm')
```

```python
text_content = nlp(content)
```

```python
text_content
```

```python
[token.text for token in text_content]
```

## SpaCy Sentence Tokenization

```python
for num,sent in enumerate(text_content.sents):
    print(f'{num}: {sent}')
```

## Word Tokenization without using NLTK/spacy 

```python
content.split()
```

## Sentence Tokenization without using NLTK/spacy 

```python
[s.strip() for s in re.split(r'(?<=[.!?])\s+', content) if s.strip()]
```
