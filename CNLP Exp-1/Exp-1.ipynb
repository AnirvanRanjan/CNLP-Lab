#  Experiment-1: Data Analysis


## Experiment-1.1: Read the employee dataset and display the average salary of each department using a bar chart.


```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("/Users/anirvanranjan/Downloads/CNLP-Lab/CNLP Exp-1/Exp-1.ipynb")
```

```python
df.info()
df.describe()
```

```python
df
```

```python
sal_by_dep = df.groupby("Department")["Salary"].mean()
```

```python
sal_by_dep.plot(kind="bar")
plt.show()
```

## Experiment 1.2: Display the number of employees working in each department.


```python
cnt_each_dept = df.groupby("Department")["Department"].count()
```

```python
cnt_each_dept
```

```python
cnt_each_dept.plot(kind="bar", color = "orange")
plt.show()
```

## Experiment 1.3:Display the percentage of male and female employees using a pie chart


```python
gender_pie = df.groupby("Gender")["Gender"].count()
```

```python
gender_pie.plot(kind="pie")
plt.show()
```

## Experiment 1.4: Draw a histogram showing salary distribution.


```python
sal = df["Salary"]
```

```python
sal.plot(kind="hist")
plt.show()
```

## Experiment 1.5: Draw a scatter plot showing the relationship between experience and salary.


```python
df.plot.scatter( x='Experience_Years', y='Salary', title='Relationship between Experience and Salary',color='blue')
plt.show()
```

## Experiment 1.6: Display the top 10 employees with the highest salary.


```python
top_10_sal = df.sort_values(by= "Salary", ascending=False)
```

```python
top_10_sal.head(10)
```

## Experiment 1.7: Find the highest salary in every department.


```python
highest_in_each_dept = df.groupby("Department")["Salary"].max()
```

```python
highest_in_each_dept.plot(kind="bar",color = "green")
plt.show()
```

## Experiment 1.8: Display employees whose salary is greater than the overall average salary.


```python
more_than_mean = df[df["Salary"] > df["Salary"].mean()]
```

```python
more_than_mean
```

## Experiment 1.9: Calculate the average years of experience in each department.


```python
avg_yrs_exp = df.groupby("Department")["Experience_Years"].mean()
```

```python
avg_yrs_exp
```
