# Experiment-4: Term Frequency and NER

## 4.1. To perform Term-Frequency Analysis and Named Entity Recognition on a given text using any toolkit.

### Calculate the frequency of each word and save the output in CSV format (Term, Frequency)

```python
import re
from collections import Counter
import pandas as pd
import spacy
import string
```

```python
file_path = "4.1_4.2_input.txt"

with open(file_path, "r", encoding="utf-8") as f:
    text = f.read()
```

```python
words = re.findall(r"\b\w+\b", text.lower())
word_counts = Counter(words)
```

### Identify the top 10 most frequent terms and display their corresponding frequencies.

```python
df_tf = pd.DataFrame(word_counts.items(), columns=["Term", "Frequency"])
df_tf = df_tf.sort_values(by="Frequency", ascending=False).reset_index(
    drop=True
)
df_tf.to_csv("term_frequencies.csv", index=False)
```

```python
top_10 = df_tf.head(10)
print("=== Top 10 Most Frequent Terms ===")
print(top_10.to_string(index=False))
```

### Perform Named Entity Recognition and display the entities found.

```python
nlp = spacy.load("en_core_web_sm")
doc = nlp(text)

df_ner = pd.DataFrame(
    [(ent.text, ent.label_, spacy.explain(ent.label_)) for ent in doc.ents],
    columns=["Entity", "Label", "Description"],
)
print(f"Total named entities found: {len(doc.ents)}")
df_ner.head(20)
```

```python
ner_counts = Counter(ent.label_ for ent in doc.ents)

print("=== Named Entity Count by Type ===")
for label, count in ner_counts.most_common():
    print(f"{label:10} {count:4}   {spacy.explain(label)}")
```

## 4.2. To perform Term-Frequency Analysis on a given text without using any toolkit.

### Calculate the frequency of each word and save the output in CSV format (Term, Frequency)

```python
file_path = "4.1_4.2_input.txt"

with open(file_path, "r", encoding="utf-8") as f:
    text = f.read()
```

```python
split_arr = text.split()
```

```python
hash_map = dict()

for x in range(len(split_arr)):
    term = split_arr[x].strip(string.punctuation).lower()
    if not term:  # token was pure punctuation
        continue
    split_arr[x] = term
    hash_map[term] = hash_map.get(term, 0) + 1
```

```python
df_tf_2 = pd.DataFrame(hash_map.items(), columns=["Term", "Frequency"])
df_tf_2 = df_tf_2.sort_values(by="Frequency", ascending=False).reset_index(
    drop=True
)
df_tf_2.to_csv("term_frequencies_without_toolkit.csv", index=False)
```

# 4.3. Write a Python program without using NLTK, spaCy, or any other NLP toolkit to:

### 1. Read multiple text documents.

```python
doc1 = "Natural language processing is a field of artificial intelligence."
doc2 = "Natural language processing helps computers understand human language."
doc3 = "Machine learning is an important part of artificial intelligence."
```

### 2. Convert text to lowercase.

```python
docs = [doc1, doc2, doc3]

# Step 2: convert every document to lowercase
docs_lower = [d.lower() for d in docs]
docs_lower
```

### 3. Remove punctuation.

```python
# Step 3: remove punctuation from every document
translator = str.maketrans("", "", string.punctuation)
docs_nopunct = [d.translate(translator) for d in docs_lower]
docs_nopunct
```

### 4. Tokenize the documents.

```python
# Step 4: tokenize each document (split on whitespace, drop empty tokens)
tokenized_docs = [[tok for tok in d.split() if tok] for d in docs_nopunct]

for i in range(len(tokenized_docs)):
    print(f"Doc{i + 1} ({len(tokenized_docs[i])} tokens): {tokenized_docs[i]}")
```

### 5. Calculate TF for every term.

```python
tf_rows = []

for i in range(len(tokenized_docs)):
    doc_len = len(tokenized_docs[i])
    term_counts = dict()
    for token in tokenized_docs[i]:
        term_counts[token] = term_counts.get(token, 0) + 1
    for term, count in term_counts.items():
        tf_rows.append((f"Doc{i + 1}", term, count, doc_len, count / doc_len))

df_tf_docs = pd.DataFrame(
    tf_rows, columns=["Document", "Term", "Count", "DocLength", "TF"]
)
df_tf_docs
```

### 6. Calculate DF for every term.

```python
import math

df_counts = dict()
for tokens in tokenized_docs:
    unique_terms = set(tokens)
    for term in unique_terms:
        df_counts[term] = df_counts.get(term, 0) + 1

df_df = (
    pd.DataFrame(df_counts.items(), columns=["Term", "DF"])
    .sort_values("Term")
    .reset_index(drop=True)
)
df_df
```

### 7. Calculate IDF using the logarithmic formula.

```python
num_docs = len(tokenized_docs)
df_df["IDF"] = df_df["DF"].apply(lambda d: math.log(num_docs / d))
df_df
```

### 8. Calculate TF-IDF for every term in every document.

```python
df_tfidf = df_tf_docs.merge(df_df, on="Term")
df_tfidf["TF-IDF"] = df_tfidf["TF"] * df_tfidf["IDF"]
df_tfidf
```

### 9. Display Term, TF, DF, IDF, and TF-IDF at each step.

```python
df_display = df_tfidf[["Document", "Term", "TF", "DF", "IDF", "TF-IDF"]]
df_display
```

### 10. Identify the top 10 terms with the highest TF-IDF score for each document.

```python
for doc in df_tfidf["Document"].unique():
    top_10_doc = (
        df_tfidf[df_tfidf["Document"] == doc]
        .sort_values(by="TF-IDF", ascending=False)
        .head(10)
        .reset_index(drop=True)
    )
    print(f"=== Top 10 Terms by TF-IDF for {doc} ===")
    print(top_10_doc[["Term", "TF", "DF", "IDF", "TF-IDF"]].to_string(index=False))
    print()
```
